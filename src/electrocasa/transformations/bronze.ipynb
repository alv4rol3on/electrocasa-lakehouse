```python
%sql
CREATE TABLE IF NOT EXISTS electrocasa.bronze.tracking_envios
AS
SELECT
    tracking_id,
    pedido_id,
    courier,
    estado_entrega,
    sucursal_origen,
    fecha_actualizacion,
    current_timestamp() AS fecha_ingestion,
    'azure_sql_tracking' AS sistema_origen,
    'tracking_inicial' AS batch_id
FROM trackingenvios_catalog.dbo.trackingenvios;
```

```python
%sql
SELECT
    COUNT(*) AS total_filas,
    COUNT(DISTINCT tracking_id) AS tracking_unicos,
    COUNT(DISTINCT pedido_id) AS pedidos_unicos
FROM electrocasa.bronze.tracking_envios;

DESCRIBE TABLE electrocasa.bronze.tracking_envios;
```

```python
%sql
CREATE TABLE IF NOT EXISTS electrocasa.bronze.catalogo_bronze (
    categoria STRING,
    marca STRING,
    nombre_producto STRING,
    precio_lista DOUBLE,
    producto_id STRING,
    _rescued_data STRING,
    fecha_ingestion TIMESTAMP,
    sistema_origen STRING,
    batch_id STRING
);
```

```python
%sql
COPY INTO electrocasa.bronze.catalogo_bronze
FROM (
    SELECT
        categoria,
        marca,
        nombre_producto,
        try_cast(precio_lista AS DOUBLE) AS precio_lista,
        producto_id,
        CAST(NULL AS STRING) AS _rescued_data,
        current_timestamp() AS fecha_ingestion,
        'catalogo_productos.json' AS sistema_origen,
        'catalogo_inicial' AS batch_id
    FROM '/Volumes/electrocasa/bronze/landing/catalogo/'
)
FILEFORMAT = JSON
FORMAT_OPTIONS (
    'multiLine' = 'true'
)
COPY_OPTIONS (
    'mergeSchema' = 'true'
);
```
